### *数组运算*

```python
import numpy as np
a = np.array([3, 4, 5])
b = np.array([1, 2, 3])
c = np.array([1, 3 ,2])
```

```python
# +, -, *, /, //, **, %取模, -取反 （对数组相应的各个元素进行运算）
a+b, np.add(a, b),   a-b, np.subtract(a, b),   a*b, np.multiply(a, b),   a/b, np.divide(a, b),   a//b, np.floor_divide(a, b),   a**b, np.power(a, b),   a%b, np.remainder(a, b),   np.negative(a)
```

```python
# ==, !=, <, <=, >, >= （对数组相应的各个元素进行比较，结果返回bool值）
b==c, np.equal(b, c),   b!=c, np.not_equal(b, c),   b<c, np.less(b, c),   b<=c, np.less_equal(b, c),   b>c, np.greater(b, c),   b>=c, np.greater_equal(b, c)
```

```python
# 矩阵运算（矩阵乘法Dot product）
A= np.array([[1, 2], [3, 4]])
B= np.array([[2, 0], [1, 2]])
A @ B, np.dot(A, B), np.matmul(A, B)
```

```python
# 6
from numpy.typing import NDArray # 引入数组类型

logits = np.array([2.0, 1.0, 0.1])
def Softmax(arr:NDArray) -> NDArray:
    exp = np.exp(arr)
    p = exp / np.sum(exp)
    return p

Softmax(logits)
```

```python
# 7
# 深度强化学习DQN中，贝尔曼更新公式：q_target = R + γ * (1 - done) * maxQ(S', α')
Rewards = np.array([1.0, 0.0, 10.0])
dones = np.array([0, 0, 1])
γ = 0.99
next_q_values = np.array([
    [1.0, 2.0, 0.5, 0.1],
    [0.2, 0.1, 3.0, 0.5],
    [5.0, 1.0, 2.0, 4.0],
])

q_target = Rewards + γ * (1 - dones) * np.max(next_q_values, axis=1)
q_target
```

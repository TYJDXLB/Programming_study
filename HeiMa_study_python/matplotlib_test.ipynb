## 该文件用于学习matplotlib
matplotlib是一个数据可视化开源python库，可以创建静态，动态，交互式的图表(使用前先安装matplotlib)  

```python
import matplotlib.pyplot as plt
```

```python
import random


# 中文显示问题
plt.rcParams["font.sans-serif"] = ["SimHei"] # 将中文字体设置为黑体
# 绘制折线图,上下数据量应该对应
x= [1, 2, 3, 4, 5, 6, 7]
y_zk= [random.randint(9, 25) for i in x]
y_qd= [random.randint(5, 20) for i in x]

plt.figure(figsize=(10, 5)) # 设置画布大小

plt.plot(x, y_zk, label="y_zk") # 折线图
plt.plot(x, y_qd, label="y_zk") # 多个折线图直接再画一个就行

plt.title("各地气温变化") # 标题
plt.xlabel("星期") # x轴标签
plt.ylabel("温度") # y轴标签
plt.grid() # 添加网格
plt.legend(loc="upper right") # 添加图例

plt.xticks(x) # 设置x轴刻度
plt.yticks(range(0, 30, 5))



plt.show() # 显示图表
```

```python
from matplotlib.axes import Axes


# 中文显示问题
plt.rcParams["font.sans-serif"] = ["SimHei"] # 将中文字体设置为黑体
# 创建子图在一个画布上展示多个图表（figure：画布对象；axes：子图数组）
figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 6), dpi=100) # nrows几行，ncols几列，figsize尺寸，dpi分辨率

# 图1
x = [1, 2, 3, 4, 5]
y = [random.randint(3, 10) for i in x]

axes1:Axes = axes[0] # 拿到第一个图:Axes相当于加一个类型注释
axes1.bar(x, y, color="g", width=0.5) # color：颜色；width：柱子宽度
axes1.set_title("柱状图", fontsize=18) # 子图的属性相当于统一加一个set_

# 图2
x = [1, 2, 3, 4, 5]
y = [random.randint(3, 10) for i in x]

axes2:Axes = axes[1] # 拿到第一个图:Axes相当于加一个类型注释
axes2.pie(y, labels=x, autopct="%1.1f%%") # color：颜色；width：柱子宽度 1.1f%:小数点前最少1位，小数点后最多1位，要放在% %中进行转义
axes2.set_title("饼状图", fontsize=18) # 子图的属性相当于统一加一个set_
axes2.legend(loc="lower center", ncol=3, bbox_to_anchor=(0, 0)) # ncol图例行数，bbox_to_anchor相对于标准怎么移动

# 保存图片
plt.savefig("resource/matplolib_save_test.png")
```

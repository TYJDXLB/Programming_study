## 该文件用于学习pandas
pandas是一个功能强大的结构化数据分析工具，底层基于Numpy库。(使用前先安装pandas)  
核心包含：
* DataFrame(类似表格)
* Series(类似表格中的一列)

```python
# 导入pandas包
import pandas as pd
```

```python
# 构造DataFrame创建数据集
# 列表嵌套字典构造
df_1 = pd.DataFrame([
    {"姓名": "小王", "年龄": 10, "性别": "男"},
    {"姓名": "小李", "年龄": 11, "性别": "女"},
    {"姓名": "小张", "年龄": 12, "性别": "女"},
    {"姓名": "小严", "年龄": 13, "性别": "男"}
])
# 字典嵌套列表构造
df_2 = pd.DataFrame({
    "姓名": ["小赵", "小钱", "小孙", "小柯"],
    "年龄": [14, 15, 16, 17],
    "性别": ["女", "男", "女", "男"]
})
# 列表嵌套元组构造
df_3 = pd.DataFrame([
    ("小周", 18, "男"),
    ("小吴", 19, "女"),
    ("小郑", 20, "男"),
    ("小黄", 21, "女"),
    ], columns=["姓名", "年龄", "性别"]
)
# 列表嵌套列表构造
df_4 = pd.DataFrame([
    ["小魏", 22, "女"],
    ["小甘", 23, "男"],
    ["小柏", 24, "男"],
    ["小陈", 25, "女"]
    ], columns=["姓名", "年龄", "性别"]
)
df_1, df_2, df_3, df_4
```

```python
# 统计信息，获取年龄最大值，最小值，平均年龄    拿到Series(类似表格中的一列)， Series也带有索引
df_1["年龄"], f"年龄最大：{df_1["年龄"].max()}，年龄最小：{df_1["年龄"].min()}，平均年龄：{df_1["年龄"].mean():.1f}"
```

```python
# 相关属性
df_1.index.tolist() # 索引
df_1.columns.tolist() # 列名
df_1.values.tolist() # 值
df_1.size # 元数据个数，行X列
df_1.dtypes # 数据类型
df_1.shape # 维度（行，列）
```

```python
df_shop = pd.read_csv("resource/商品列表.csv", usecols=["商品ID", "商品名称", "分类", "销量", "单价"])
# 配置项，显示所有数据
pd.set_option("display.max_rows", None)
# 数据查询
df_shop.head(10) # 显示前10行数据
df_shop.tail(10) # 显示后10行数据
df_shop.describe() # 显示数据统计信息
df_shop.info() # 显示数据信息
df_shop.columns # 显示列名
df_shop.shape # 显示形状
```

```python
# 数据选择
# 选择列
df_shop["商品ID"] # 单列
df_shop.单价
df_shop[["商品ID", "商品名称"]] # 双列
# 选择行
df_shop.iloc[0:11:2] # 基于行号选择，语法: .iloc[start:stop:step]（不包含stop）
df_shop.loc[0:11:2] # 基于索引选择，语法: .loc[start:stop:step]（包含stop）
```

```python
# 数据过滤
df_shop[df_shop["销量"] > 100] # 选择销量大于8的
df_shop[df_shop["分类"].isin(["智能家电", "运动户外"])] # 选择一列的某几个标签
df_shop[df_shop["销量"].between(100, 150, inclusive="left")] # 介于二者之间默认包含边界inclusive=只包含哪边
df_shop[(df_shop["单价"] > 100) & (df_shop["销量"].between(100, 150, inclusive="left"))] # 多条件和：& ；或：|
```

```python
# 数据清洗：发现并纠正数据中可能存在的错误，包括处理缺失值，重复值，异常值，统一数据格式
df_shop_2 = pd.read_csv("resource/商品列表.csv")
pd.set_option("display.min_rows", 90) # 显示所有行
# 数据处理
# 缺失数值处理
df_shop_2.isnull() # 查看缺失值
df_shop_2.dropna() # 删除缺失值所在的行，删完之后返回一个新的df需要赋值拿到
df_shop_2.dropna(axis=1) # 删除缺失值所在的列
df_shop_2.fillna("haha") # 填充缺失值
df_shop_2.ffill() # 填充缺失值为上一行内容
df_shop_2.bfill() # 填充缺失值为下一行内容
# 重复值处理
df_shop_2.duplicated() # 查看重复值（所有的列的数据都一样）
df_shop_2.duplicated(subset=['商品ID']) # 查看重复值（指定列的数据都一样）
df_shop_2.drop_duplicates(subset=['商品ID'], keep="first") # 删除指定行的全部重复行,keep='first'保留第一行，keep='last'保留最后一行，keep=false，不保留
# 异常值处理
df_shop_2[df_shop_2["库存"] < 0] # 查询异常值
df_shop_2.drop(df_shop_2[df_shop_2["库存"] < 0].index) # 删除异常值
df_shop_2["库存"] = df_shop_2["库存"].abs() # 取绝对值
# 数据格式处理
# df_shop_2["上架日期"] = df_shop_2["上架日期"].replace('2025/05/18', '2025-05-18') # 对单元格进行整体替换
df_shop_2["上架日期"] = df_shop_2["上架日期"].str.replace('/', '-') # 加上.str就可以调用str的方法了
# 数据排序（多列排序的时候先排第一列再排第二列）.sort_values(["销量", "评分"])
df_shop_2.sort_values("销量", ascending=False) # ascending为False时降序，True为升序
df_shop_2.sort_values(["销量", "重量kg"], ascending=[True, False]) # 先排第一列，第一列一样的时候再按规则排第二列
# 数据分组
df_shop_2.groupby("分类").count() # 根据某一列进行分组,显示数量
df_shop_2.groupby("分类")["品牌"].count() # 根据某一列进行分组,只显示某一列的分组
df_shop_2.groupby("分类")["单价"].sum() # 统计求和sum，最小值min，最大值max，平均值mean
df_shop_2.groupby("分类")["单价"].agg(["min", "max", "mean"]) # 同时统计同列的多个信息
df_shop_2.groupby("分类").agg({"单价": "min", "库存": "max", "销量": "mean"}) # 同时统计不同列的多个信息

```

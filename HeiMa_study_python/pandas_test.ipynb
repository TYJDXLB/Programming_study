## 该文件用于学习pandas
pandas是一个功能强大的结构化数据分析工具，底层基于Numpy库。(使用前先安装pandas)  
核心包含：
* DataFrame(类似表格)
* Series(类似表格中的一列)

```python
# 导入pandas包
import pandas as pd
```

```python
# 构造DataFrame创建数据集
df = pd.DataFrame([
    {"姓名": "小王", "年龄": 18, "性别": "男"},
    {"姓名": "小李", "年龄": 19, "性别": "女"},
    {"姓名": "小赵", "年龄": 20, "性别": "女"},
    {"姓名": "小严", "年龄": 22, "性别": "男"}
])
df
```

```python
# 统计信息，获取年龄最大值，最小值，平均年龄    拿到Series(类似表格中的一列)， Series也带有索引
df["年龄"], f"年龄最大：{df["年龄"].max()}，年龄最小：{df["年龄"].min()}，平均年龄：{df["年龄"].mean():.1f}"
```

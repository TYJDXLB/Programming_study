## 该文件用于学习pandas
pandas是一个功能强大的结构化数据分析工具，底层基于Numpy库。(使用前先安装pandas)  
核心包含：
* DataFrame(类似表格)
* Series(类似表格中的一列)

```python
# 导入pandas包
import pandas as pd
```

```python
# 构造DataFrame创建数据集
# 列表嵌套字典构造
df_1 = pd.DataFrame([
    {"姓名": "小王", "年龄": 10, "性别": "男"},
    {"姓名": "小李", "年龄": 11, "性别": "女"},
    {"姓名": "小张", "年龄": 12, "性别": "女"},
    {"姓名": "小严", "年龄": 13, "性别": "男"}
])
# 字典嵌套列表构造
df_2 = pd.DataFrame({
    "姓名": ["小赵", "小钱", "小孙", "小柯"],
    "年龄": [14, 15, 16, 17],
    "性别": ["女", "男", "女", "男"]
})
# 列表嵌套元组构造
df_3 = pd.DataFrame([
    ("小周", 18, "男"),
    ("小吴", 19, "女"),
    ("小郑", 20, "男"),
    ("小黄", 21, "女"),
    ], columns=["姓名", "年龄", "性别"]
)
# 列表嵌套列表构造
df_4 = pd.DataFrame([
    ["小魏", 22, "女"],
    ["小甘", 23, "男"],
    ["小柏", 24, "男"],
    ["小陈", 25, "女"]
    ], columns=["姓名", "年龄", "性别"]
)
df_1, df_2, df_3, df_4
```

```python
# 统计信息，获取年龄最大值，最小值，平均年龄    拿到Series(类似表格中的一列)， Series也带有索引
df_1["年龄"], f"年龄最大：{df_1["年龄"].max()}，年龄最小：{df_1["年龄"].min()}，平均年龄：{df_1["年龄"].mean():.1f}"
```

```python
# 相关属性
df_1.index.tolist() # 索引
df_1.columns.tolist() # 列名
df_1.values.tolist() # 值
df_1.size # 元数据个数，行X列
df_1.dtypes # 数据类型
df_1.shape # 维度（行，列）
```

```python
# 构建Series
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series((6, 7, 8, 9, 10), index=['a', 'b', 'c', 'd', 'e'])
s3 =pd.Series({'f': 11, 'g': 12, 'h': 13, 'i': 14, 'j': 15})
s_df_1 = df_1["姓名"]
s_df_2 = df_1.性别

s1, s_df_1, s_df_2
```
